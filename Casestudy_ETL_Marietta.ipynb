Q6

```python
gdp=gdp_df.rename(columns={"Country":"country"})
literacy=literacy_df.rename(columns={"name":"country"})
print(gdp)
print(literacy)
```

Q8

```python
merged_df = pd.merge(population_df, gdp, on="country", how="inner")
merged_df = pd.merge(merged_df, internet_df, on="country", how="inner")
merged_df = pd.merge(merged_df, literacy, on="country", how="inner")

merged_df
```

Q9

```python
merged_df["Internet Penetration (%)"] = ((merged_df["internet_users"] / merged_df["population"]) * 100).round(2)

merged_df["GDP Per Capita"] = (merged_df["GDP"] / merged_df["population"]).round(2)

merged_df
```

Q10

```python
import sqlite3
conn=sqlite3.connect("country_statistics.db")
merged_df.to_sql("country_statistics", conn, if_exists="replace", index=False)
```

Q11

```python
query = "select*from country_statistics LIMIT 5"
result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT country, GDP
FROM country_statistics
ORDER BY GDP DESC
LIMIT 5
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT AVG(literacy_rate) AS average_literacy
FROM country_statistics
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT country, "Internet Penetration (%)"
FROM country_statistics
ORDER BY "Internet Penetration (%)" DESC
LIMIT 5
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT SUM(population) AS total_population
FROM country_statistics
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT country, "GDP Per Capita"
FROM country_statistics
ORDER BY "GDP Per Capita" DESC
LIMIT 1
"""

pd.read_sql(query, conn)
```
